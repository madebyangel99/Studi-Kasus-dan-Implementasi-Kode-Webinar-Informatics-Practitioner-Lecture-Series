```python
import matplotlib.pyplot as plt
import numpy as np
```

1. Data Penjualan (dalam Juta Rp)

```python
bulan = ['Jan', 'Feb', 'Mar', 'Apr', 'Mei', 'Jun']
penjualan = [5, 7, 6, 8, 12, 9]
```

2. Pengaturan Tema Warna (Identitas UNIB: Biru Tua, Kuning, Putih)

```python
bg_color = '#ffffff'
biru_tua = '#003366'
kuning_unib = '#FFCC00'
text_color = '#003366'

# Menerapkan Pre-attentive attribute: Mewarnai otomatis nilai tertinggi dengan Kuning
max_val = max(penjualan)
colors = [kuning_unib if val == max_val else biru_tua for val in penjualan]
```

3. Membuat Figure dan Axes

```python
fig, ax = plt.subplots(figsize=(10, 6))
fig.patch.set_facecolor(bg_color)
ax.set_facecolor(bg_color)

# Membuat Bar Chart
bars = ax.bar(bulan, penjualan, color=colors, width=0.6, edgecolor='white', linewidth=1.5)

# Recognition over Recall: Menampilkan label angka langsung di atas batang
for bar in bars:
    yval = bar.get_height()
    # Menggunakan warna teks kuning untuk batang yang kuning agar kontras, biru untuk yang lain
    label_color = biru_tua if yval != max_val else '#B38F00' # Kuning gelap untuk teks agar terbaca di latar putih

    ax.text(bar.get_x() + bar.get_width()/2, yval + 0.3, f'{yval}M',
            ha='center', va='bottom', color=label_color, fontweight='bold')
```

4. Etika Desain: Skala Y Jujur (Harus mulai dari 0)

```python
ax.set_ylim(bottom=0, top=14)
```

5. Mengurangi Extraneous Load: Membuang garis tepi yang tidak perlu

```python
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.spines['left'].set_visible(False)
ax.spines['bottom'].set_color(biru_tua)
ax.spines['bottom'].set_linewidth(2)

# Pengaturan Tipografi dan Grid
ax.set_title("Dashboard Penjualan UMKM", fontsize=16, fontweight='bold', color=biru_tua, pad=20)
ax.set_ylabel("Pendapatan (Juta Rp)", color=biru_tua, fontweight='bold')
ax.tick_params(axis='x', colors=biru_tua)
ax.tick_params(axis='y', colors=biru_tua, left=False)

# Grid samar agar rapi, menggunakan abu-abu sangat terang agar tidak bertabrakan dengan biru/kuning
ax.grid(axis='y', linestyle='--', alpha=0.3, color=biru_tua)

# Render visualisasi
plt.tight_layout()
plt.show()
```
